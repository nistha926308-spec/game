```python
from IPython.display import HTML, display

display(HTML("""
<div style="font-family:sans-serif;text-align:center;max-width:420px">
  <div id="score" style="font-size:18px;margin-bottom:6px">Score: 0 &nbsp;|&nbsp; Best: 0</div>
  <canvas id="game" width="400" height="400" tabindex="0"
          style="background:#111;border:3px solid #444;border-radius:6px;outline:none"></canvas>
  <div id="msg" style="margin-top:6px;color:#888">Click the game, then use arrow keys / WASD. Space = restart.</div>
  <div style="margin-top:8px">
    <button onclick="setDir(0,-1)">&#9650;</button><br>
    <button onclick="setDir(-1,0)">&#9664;</button>
    <button onclick="setDir(0,1)">&#9660;</button>
    <button onclick="setDir(1,0)">&#9654;</button>
  </div>
</div>
<script>
(function(){
  const cv = document.getElementById('game'), ctx = cv.getContext('2d');
  const N = 20, S = cv.width / N;
  let snake, dir, nextDir, food, score, best = 0, alive, timer;

  function reset(){
    snake = [{x:10,y:10},{x:9,y:10},{x:8,y:10}];
    dir = {x:1,y:0}; nextDir = dir; score = 0; alive = true;
    placeFood(); updateScore();
    document.getElementById('msg').textContent = 'Click the game, then use arrow keys / WASD. Space = restart.';
    clearInterval(timer); timer = setInterval(tick, 110); draw();
  }
  function placeFood(){
    do { food = {x:Math.floor(Math.random()*N), y:Math.floor(Math.random()*N)}; }
    while (snake.some(s => s.x===food.x && s.y===food.y));
  }
  function updateScore(){
    document.getElementById('score').innerHTML = 'Score: '+score+' &nbsp;|&nbsp; Best: '+best;
  }
  window.setDir = function(x,y){
    if (x === -dir.x && y === -dir.y) return;   // no reversing
    nextDir = {x:x, y:y};
  };
  function tick(){
    if (!alive) return;
    dir = nextDir;
    const head = {x:snake[0].x+dir.x, y:snake[0].y+dir.y};
    if (head.x<0||head.y<0||head.x>=N||head.y>=N || snake.some(s=>s.x===head.x&&s.y===head.y)){
      alive = false; clearInterval(timer);
      best = Math.max(best, score); updateScore();
      document.getElementById('msg').textContent = 'Game over! Press Space to restart.';
      draw(); return;
    }
    snake.unshift(head);
    if (head.x===food.x && head.y===food.y){ score++; best = Math.max(best, score); updateScore(); placeFood(); }
    else snake.pop();
    draw();
  }
  function draw(){
    ctx.fillStyle = '#111'; ctx.fillRect(0,0,cv.width,cv.height);
    ctx.fillStyle = '#e74c3c';
    ctx.beginPath(); ctx.arc(food.x*S+S/2, food.y*S+S/2, S/2-2, 0, Math.PI*2); ctx.fill();
    snake.forEach((s,i)=>{
      ctx.fillStyle = i===0 ? '#2ecc71' : '#27ae60';
      ctx.fillRect(s.x*S+1, s.y*S+1, S-2, S-2);
    });
    if (!alive){
      ctx.fillStyle = 'rgba(0,0,0,0.6)'; ctx.fillRect(0,0,cv.width,cv.height);
      ctx.fillStyle = '#fff'; ctx.font = '28px sans-serif'; ctx.textAlign = 'center';
      ctx.fillText('Game Over', cv.width/2, cv.height/2);
    }
  }
  cv.addEventListener('click', () => cv.focus());
  cv.addEventListener('keydown', e => {
    const k = e.key.toLowerCase();
    const map = {arrowup:[0,-1],w:[0,-1],arrowdown:[0,1],s:[0,1],arrowleft:[-1,0],a:[-1,0],arrowright:[1,0],d:[1,0]};
    if (map[k]){ setDir(...map[k]); e.preventDefault(); }
    else if (k === ' '){ reset(); e.preventDefault(); }
  });
  reset(); cv.focus();
})();
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML("""
<div style="font-family:sans-serif;text-align:center;max-width:420px">
  <div id="score" style="font-size:18px;margin-bottom:6px">Score: 0 &nbsp;|&nbsp; Best: 0</div>
  <canvas id="game" width="400" height="400" tabindex="0"
          style="background:#111;border:3px solid #444;border-radius:6px;outline:none"></canvas>
  <div id="msg" style="margin-top:6px;color:#888">Click the game, then use arrow keys / WASD. Space = restart.</div>
  <div style="margin-top:4px;color:#888;font-size:13px">&#127822; 1 pt &nbsp; &#127820; 2 pts &nbsp; &#127815; 3 pts &nbsp; &#127826; 5 pts (rare)</div>
  <div style="margin-top:8px">
    <button onclick="setDir(0,-1)">&#9650;</button><br>
    <button onclick="setDir(-1,0)">&#9664;</button>
    <button onclick="setDir(0,1)">&#9660;</button>
    <button onclick="setDir(1,0)">&#9654;</button>
  </div>
</div>
<script>
(function(){
  const cv = document.getElementById('game'), ctx = cv.getContext('2d');
  const N = 20, S = cv.width / N, FRUIT_COUNT = 3;

  // emoji, points, spawn weight
  const FRUITS = [
    {e:'\\u{1F34E}', pts:1, w:50},   // apple
    {e:'\\u{1F34C}', pts:2, w:25},   // banana
    {e:'\\u{1F347}', pts:3, w:18},   // grapes
    {e:'\\u{1F352}', pts:5, w:7}     // cherries (rare)
  ];

  let snake, dir, nextDir, fruits, score, best = 0, alive, timer, pending;

  function pickFruitType(){
    const total = FRUITS.reduce((a,f)=>a+f.w,0);
    let r = Math.random()*total;
    for (const f of FRUITS){ if ((r -= f.w) < 0) return f; }
    return FRUITS[0];
  }
  function addFruit(){
    let pos;
    do { pos = {x:Math.floor(Math.random()*N), y:Math.floor(Math.random()*N)}; }
    while (snake.some(s=>s.x===pos.x&&s.y===pos.y) || fruits.some(f=>f.x===pos.x&&f.y===pos.y));
    const t = pickFruitType();
    fruits.push({x:pos.x, y:pos.y, e:t.e, pts:t.pts});
  }
  function reset(){
    snake = [{x:10,y:10},{x:9,y:10},{x:8,y:10}];
    dir = {x:1,y:0}; nextDir = dir; score = 0; alive = true; pending = 0;
    fruits = []; for (let i=0;i<FRUIT_COUNT;i++) addFruit();
    updateScore();
    document.getElementById('msg').textContent = 'Click the game, then use arrow keys / WASD. Space = restart.';
    clearInterval(timer); timer = setInterval(tick, 110); draw();
  }
  function updateScore(){
    document.getElementById('score').innerHTML = 'Score: '+score+' &nbsp;|&nbsp; Best: '+best;
  }
  window.setDir = function(x,y){
    if (x === -dir.x && y === -dir.y) return;   // no reversing
    nextDir = {x:x, y:y};
  };
  function tick(){
    if (!alive) return;
    dir = nextDir;
    const head = {x:snake[0].x+dir.x, y:snake[0].y+dir.y};
    if (head.x<0||head.y<0||head.x>=N||head.y>=N || snake.some(s=>s.x===head.x&&s.y===head.y)){
      alive = false; clearInterval(timer);
      best = Math.max(best, score); updateScore();
      document.getElementById('msg').textContent = 'Game over! Press Space to restart.';
      draw(); return;
    }
    snake.unshift(head);
    const idx = fruits.findIndex(f=>f.x===head.x&&f.y===head.y);
    if (idx !== -1){
      const f = fruits.splice(idx,1)[0];
      score += f.pts; pending += f.pts;
      best = Math.max(best, score); updateScore();
      addFruit();
    }
    if (pending > 0) pending--; else snake.pop();   // grow by fruit value
    draw();
  }
  function draw(){
    ctx.fillStyle = '#111'; ctx.fillRect(0,0,cv.width,cv.height);

    // fruits
    ctx.font = (S-2)+'px serif'; ctx.textAlign = 'center'; ctx.textBaseline = 'middle';
    fruits.forEach(f => ctx.fillText(f.e, f.x*S+S/2, f.y*S+S/2+2));

    // snake: cyan head, gradient body fading to purple
    snake.forEach((s,i)=>{
      ctx.fillStyle = i===0 ? '#00e5ff' : 'hsl('+(190+Math.min(i*6,100))+',80%,55%)';
      ctx.beginPath();
      ctx.roundRect ? ctx.roundRect(s.x*S+1, s.y*S+1, S-2, S-2, 5) : ctx.rect(s.x*S+1, s.y*S+1, S-2, S-2);
      ctx.fill();
    });
    // eyes
    const h = snake[0];
    ctx.fillStyle = '#fff';
    const ex = dir.y!==0 ? [5,S-5] : [dir.x>0?S-6:6, dir.x>0?S-6:6];
    const ey = dir.y!==0 ? [dir.y>0?S-6:6, dir.y>0?S-6:6] : [5,S-5];
    for (let k=0;k<2;k++){ ctx.beginPath(); ctx.arc(h.x*S+ex[k], h.y*S+ey[k], 2.2, 0, Math.PI*2); ctx.fill(); }

    if (!alive){
      ctx.fillStyle = 'rgba(0,0,0,0.6)'; ctx.fillRect(0,0,cv.width,cv.height);
      ctx.fillStyle = '#fff'; ctx.font = '28px sans-serif'; ctx.textAlign = 'center';
      ctx.fillText('Game Over', cv.width/2, cv.height/2);
    }
  }
  cv.addEventListener('click', () => cv.focus());
  cv.addEventListener('keydown', e => {
    const k = e.key.toLowerCase();
    const map = {arrowup:[0,-1],w:[0,-1],arrowdown:[0,1],s:[0,1],arrowleft:[-1,0],a:[-1,0],arrowright:[1,0],d:[1,0]};
    if (map[k]){ setDir(...map[k]); e.preventDefault(); }
    else if (k === ' '){ reset(); e.preventDefault(); }
  });
  reset(); cv.focus();
})();
</script>
"""))
```
